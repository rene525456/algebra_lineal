### Aplicación Interactiva para Graficar Ecuaciones Lineales

Esta aplicación te permite visualizar una ecuación lineal de la forma general:

$$Ax + By = C$$

O de la forma pendiente-intersección (si despejamos $y$ cuando $B \neq 0$):

$$y = -\frac{A}{B}x + \frac{C}{B}$$

Usa los deslizadores para cambiar los valores de $A$, $B$ y $C$.

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def graficar_lineal(A, B, C):
    fig, ax = plt.subplots(figsize=(8, 8))

    # Configurar límites y cuadrícula
    ax.set_xlim(-10, 10)
    ax.set_ylim(-10, 10)
    ax.axhline(0, color='black', linewidth=1.2)
    ax.axvline(0, color='black', linewidth=1.2)
    ax.grid(True, which='both', linestyle='--', alpha=0.5)

    # Título dinámico
    formula = f"{A}x + {B}y = {C}"
    # Reemplazar dobles signos para mejorar legibilidad
    formula = formula.replace("+ -", "- ").replace("- -", "+ ")
    ax.set_title(f"Ecuación: {formula}", fontsize=14, pad=15)
    ax.set_xlabel("Eje X", fontsize=12)
    ax.set_ylabel("Eje Y", fontsize=12)

    # Casos para graficar e imprimir valores
    if A == 0 and B == 0:
        # Caso inválido 0 = C
        status_text = f"¡Inválido! 0 = {C}" if C != 0 else "Identidad: 0 = 0"
        ax.text(0, 0, status_text, fontsize=12, color='red', ha='center', va='center',
                bbox=dict(facecolor='white', alpha=0.8, edgecolor='red'))
        print("No hay valores definidos para x y y en una ecuación inválida.")
    elif B == 0:
        # Línea vertical: x = C / A
        x_val = C / A
        ax.axvline(x_val, color='blue', linewidth=2.5, label=f'x = {x_val:.2f}')
        ax.legend(loc='upper right')

        print(f"Línea vertical en x = {x_val:.2f}")
        print("Valores de ejemplo (x, y):")
        for y_val in range(-5, 6, 2):
            print(f"({x_val:.2f}, {y_val:.2f})")
    else:
        # Línea estándar o horizontal: y = (C - Ax) / B
        x = np.linspace(-10, 10, 400)
        y = (C - A * x) / B

        pendiente = -A / B
        intercepto = C / B
        label_recta = f'y = {pendiente:.2f}x + {intercepto:.2f}'.replace("+ -", "- ")

        ax.plot(x, y, color='blue', linewidth=2.5, label=label_recta)
        ax.legend(loc='upper right')

        print(f"Ecuación explícita: {label_recta}")
        print("Valores de ejemplo (x, y):")
        # Mostrar algunos puntos clave
        x_muestras = np.linspace(-10, 10, 11)
        for x_m in x_muestras:
            y_m = (C - A * x_m) / B
            print(f"x: {x_m:6.2f}  ->  y: {y_m:6.2f}")

    plt.show()

# Crear controles deslizantes interactivos con valores predeterminados para 2x + 3y = 6

print("Graficar ecuaciones lineales de la forma Ax + By = C")
print("Jordy Pineda")
graficar_lineal(5, 5, 1)
```
