```python
import numpy as np
import matplotlib.pyplot as plt

def resolver_y_graficar_sustitucion(a1, b1, c1, a2, b2, c2):
    """
    Resuelve un sistema de 2 ecuaciones lineales con 2 incógnitas por el método de sustitución
    y grafica el resultado indicando el punto de intersección.
    Ecuación 1: a1*x + b1*y = c1
    Ecuación 2: a2*x + b2*y = c2
    """
    print("--- MÉTODO DE SUSTITUCIÓN ---")
    print(f"Sistema de ecuaciones:")
    print(f"(1) {a1}x + ({b1})y = {c1}")
    print(f"(2) {a2}x + ({b2})y = {c2}\n")

    # Verificar si el sistema tiene solución única (determinante no nulo)
    det = a1 * b2 - a2 * b1
    if det == 0:
        if (a1 * c2 == a2 * c1) and (b1 * c2 == b2 * c1):
            print("El sistema tiene infinitas soluciones (es indeterminado).")
        else:
            print("El sistema no tiene solución (es incompatible).")
        return None

    # Paso 1: Despejar 'x' de la ecuación (1)
    if a1 != 0:
        print(f"Paso 1: Despejamos 'x' de la ecuación (1):")
        print(f"  x = ({c1} - ({b1})y) / {a1}\n")

        # Paso 2: Sustituir 'x' en la ecuación (2)
        print(f"Paso 2: Sustituimos 'x' en la ecuación (2):")
        print(f"  {a2} * [({c1} - ({b1})y) / {a1}] + ({b2})y = {c2}")

        factor_y = b2 - (a2 * b1) / a1
        termino_independiente = c2 - (a2 * c1) / a1
        print(f"  y * ({factor_y}) = {termino_independiente}")

        # Paso 3: Resolver para 'y'
        y = termino_independiente / factor_y
        print(f"  y = {y}\n")

        # Paso 4: Sustituir 'y' en el despeje de 'x'
        print(f"Paso 3: Sustituimos 'y' en la expresión despejada de 'x':")
        x = (c1 - b1 * y) / a1
        print(f"  x = ({c1} - ({b1})*({y})) / {a1}")
        print(f"  x = {x}\n")

    else:
        print(f"Paso 1: Dado que x no está en la ec (1) (a1=0), despejamos 'y' de (1):")
        y = c1 / b1
        print(f"  y = {c1} / {b1} = {y}\n")

        print(f"Paso 2: Sustituimos 'y' en la ecuación (2):")
        print(f"  {a2}x + ({b2})*({y}) = {c2}")

        print(f"Paso 3: Resolvemos para 'x':")
        x = (c2 - b2 * y) / a2
        print(f"  {a2}x = {c2 - b2 * y}")
        print(f"  x = {x}\n")

    print("--- RESULTADO FINAL ---")
    print(f"x = {x}")
    print(f"y = {y}\n")

    # --- Gráfico de la Solución ---
    # Definimos un rango de x alrededor de la solución encontrada
    x_vals = np.linspace(x - 10, x + 10, 400)

    # Despejamos 'y' de ambas ecuaciones para graficar: y = (c - a*x) / b
    # Si b es 0, la recta es vertical x = c/a
    plt.figure(figsize=(8, 6))

    # Graficar Ecuación 1
    if b1 != 0:
        y_vals1 = (c1 - a1 * x_vals) / b1
        plt.plot(x_vals, y_vals1, label=f"{a1}x + {b1}y = {c1}", color="blue")
    else:
        plt.axvline(x=c1/a1, label=f"{a1}x = {c1}", color="blue", linestyle="--")

    # Graficar Ecuación 2
    if b2 != 0:
        y_vals2 = (c2 - a2 * x_vals) / b2
        plt.plot(x_vals, y_vals2, label=f"{a2}x + {b2}y = {c2}", color="green")
    else:
        plt.axvline(x=c2/a2, label=f"{a2}x = {c2}", color="green", linestyle="--")

    # Marcar el punto de intersección (solución)
    plt.scatter(x, y, color="red", zorder=5, s=100, label=f"Solución: ({x:.2f}, {y:.2f})")

    # Detalles del gráfico
    plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
    plt.axvline(0, color='black', linewidth=0.5, linestyle='--')
    plt.title("Representación Gráfica del Sistema de Ecuaciones")
    plt.xlabel("Eje X")
    plt.ylabel("Eje Y")
    plt.legend()
    plt.grid(True, linestyle=":", alpha=0.6)
    plt.show()

    return x, y

# Resolver y graficar el sistema ejemplo
x_sol, y_sol = resolver_y_graficar_sustitucion(a1=2, b1=1, c1=1, a2=3, b2=4, c2=14)
```
